This notebook demonstrates loading, inspecting, cleaning, filtering, grouping, and summarizing a dataset using pandas.

```python
import pandas as pd
import numpy as np
```

## 1. Create and inspect a dataset

```python
data = {
    'name': ['Asha', 'Ravi', 'Neha', 'Kabir', 'Meera', 'Ishaan', 'Sara', 'Dev'],
    'city': ['Delhi', 'Mumbai', 'Pune', 'Delhi', 'Pune', 'Mumbai', 'Delhi', 'Pune'],
    'age': [18, 19, 17, 20, 18, 21, 19, 18],
    'math_marks': [78, 85, 91, 67, 88, 73, np.nan, 95],
    'science_marks': [82, 79, 94, 72, np.nan, 81, 89, 96]
}
df = pd.DataFrame(data)
df.head()
```

```python
print('Shape:', df.shape)
print('\nDataFrame information:')
df.info()
print('\nSummary statistics:')
display(df.describe(numeric_only=True))
```

## 2. Check missing values

```python
print(df.isnull().sum())
```

## 3. Clean missing values

```python
df_clean = df.copy()
df_clean['math_marks'] = df_clean['math_marks'].fillna(df_clean['math_marks'].mean())
df_clean['science_marks'] = df_clean['science_marks'].fillna(df_clean['science_marks'].mean())
print('Missing values after cleaning:')
print(df_clean.isnull().sum())
display(df_clean)
```

## 4. Create total and average scores

```python
df_clean['total_marks'] = df_clean['math_marks'] + df_clean['science_marks']
df_clean['average_marks'] = df_clean[['math_marks', 'science_marks']].mean(axis=1)
display(df_clean[['name', 'math_marks', 'science_marks', 'total_marks', 'average_marks']])
```

## 5. Filter students with average marks of at least 85

```python
high_performers = df_clean[df_clean['average_marks'] >= 85]
display(high_performers[['name', 'city', 'average_marks']])
```

## 6. Group by city and calculate average marks

```python
city_summary = df_clean.groupby('city')[['math_marks', 'science_marks', 'average_marks']].mean().round(2)
display(city_summary)
```

## 7. Sort students by average marks

```python
ranked_students = df_clean.sort_values('average_marks', ascending=False)
display(ranked_students[['name', 'city', 'average_marks']])
```

## 8. Final findings

- Missing numeric values were filled using the column mean to retain all students.
- New features (`total_marks` and `average_marks`) were created.
- Students were filtered by average score and ranked.
- City-level averages were calculated using `groupby()`.
