## Valid Sudoku

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])

rows_valid = all(len(np.unique(row)) == 9 for row in s)
columns_valid = all(len(np.unique(col)) == 9 for col in s.T)
boxes_valid = all(
    len(np.unique(s[i:i + 3, j:j + 3])) == 9
    for i in range(0, 9, 3)
    for j in range(0, 9, 3)
)

print('Rows valid:', rows_valid)
print('Columns valid:', columns_valid)
print('3x3 boxes valid:', boxes_valid)
print('Valid Sudoku:', rows_valid and columns_valid and boxes_valid)
```

## General Questions

Columns: Age, Math Marks, Science Marks

```python
data = np.array([
    [18, 85, 78],
    [19, 92, 88],
    [17, 76, 95],
    [18, 65, 70],
    [20, 90, 85]
])

print(data)
```

### 1. Get the shape of the matrix

```python
print(data.shape)
```

### 2. Find the average age of students

```python
print(np.mean(data[:, 0]))
```

### 3. Extract Math marks of all students

```python
print(data[:, 1])
```

### 4. Find the highest Science mark

```python
print(np.max(data[:, 2]))
```

### 5. Get details of the student who scored more than 90 in Math

```python
print(data[data[:, 1] > 90])
```

### 6. Increase Math marks of all students by 5

```python
data[:, 1] += 5
print(data)
```

### 7. Find how many students are younger than 19

```python
print(np.sum(data[:, 0] < 19))
```

### 8. Calculate the average marks in each subject

```python
print(np.mean(data[:, 1:], axis=0))
```

### 9. Get students who scored at least 80 in both subjects

```python
print(data[(data[:, 1] >= 80) & (data[:, 2] >= 80)])
```

### 10. Replace all Science marks below 75 with 0

```python
data[data[:, 2] < 75, 2] = 0
print(data)
```
